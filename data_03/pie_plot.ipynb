```python
import  matplotlib.pyplot as plt
```

```python
days = ["sat" , "sun" , "mon" , "tue"  , "wed" , "thu" , "fri"]
company_1 = [99 , 104 , 110 , 115 , 120 , 100 , 105]
```

```python
plt.pie( company_1 , labels = days)

plt.show()
```

```python
plt.pie(company_1 , autopct = "%1.1f%%")

plt.show()
```
