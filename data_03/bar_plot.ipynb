```python
import  matplotlib.pyplot as plt
```

```python
days = ["sat" , "sun" , "mon" , "tue"  , "wed" , "thu" , "fri"]
company_1 = [99 , 104 , 110 , 115 , 120 , 100 , 105]
```

```python
#Bar Plot
plt.bar(days , company_1 , color = "red" , label = "company_1")

plt.xlabel("days")
plt.ylabel("company_1")

plt.legend()
plt.show()
```

```python
#Horizontal Bar Plot
plt.barh(days , company_1 , label = " company_1")

plt.legend()
plt.show()
```
