```python
import matplotlib.pyplot as plt
```

```python
days = ["sat" , "sun" , "mon" , "tue" , "wed" , "thu" , "fri"]
company_1 = [100 , 107 , 101 , 99 , 106 , 117 , 124]
```

```python
plt.scatter(days , company_1)
plt.show()
```

```python
plt.scatter(days , company_1)

plt.ylim([0 , 150])
plt.show()
```

```python
plt.scatter(days , company_1 , color = "black")
plt.show()
```

```python
plt.scatter(days , company_1  , s = 200)
plt.show()
```

```python
plt.scatter(days , company_1 , color = "red" , s = 250 , alpha = 0.5)
plt.show()
```
