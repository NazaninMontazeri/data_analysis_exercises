```python
import matplotlib.pyplot as plt
```

```python
days = ["sat" , "sun" , "mon" , "tue"  , "wed" , "thu" , "fri"]
company_1 = [99 , 104 , 110 , 115 , 120 , 100 , 105]
```

```python
plt.plot(days , company_1)

plt.ylim([90 , 130])
plt.show()
```

```python
plt.plot(days , company_1 , marker = "p")

plt.show()
```

```python
plt.plot(days , company_1 , marker = "p" , color = "g")

plt.show()
```

```python
plt.plot(days , company_1 , marker = "o" , color = "b" , markerfacecolor = "red")

plt.show()
```

```python
plt.plot(days , company_1 , marker = "o" , color = "b" , markerfacecolor = "red" , markeredgecolor = "g")

plt.show()
```
